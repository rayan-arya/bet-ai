```python
import requests, json

SITE_SUMMARY = "https://site.web.api.espn.com/apis/site/v2/sports/football/nfl/summary"

def j(url):
    r = requests.get(url, timeout=20)
    r.raise_for_status()
    return r.json()

def get_summary(event_id: str):
    return j(f"{SITE_SUMMARY}?event={event_id}")

# Example: take the first game from your 2025-10-19 pull
event_id = "401772635"   # <- replace with any event id you have
data = get_summary(event_id)

# Pretty-print the whole JSON
print(json.dumps(data, indent=2)[:2000])  # print first ~2000 chars so your console isn't flooded

# Peek common sections
print("Top-level keys:", list(data.keys()))
print("Boxscore keys:", list(data.get("boxscore", {}).keys()))
print("Competitions keys:", list(data.get("header", {}).get("competitions", [{}])[0].keys()))

# Optional: save to a file
with open(f"espn_summary_{event_id}.json", "w") as f:
    json.dump(data, f, indent=2)
print(f"Saved espn_summary_{event_id}.json")
```
